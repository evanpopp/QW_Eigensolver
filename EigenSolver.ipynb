```python
## 1D Solver ##

%load_ext autoreload
%autoreload 2

from qmsolve import Hamiltonian, SingleParticle, init_visualization,Å
from my_libs import Constants as const
from my_libs import QW_Lib as qwlib

## parameters
bias = 2            # applied voltage/bias to the quantum well #
resolution = 1000   # horizontal resolution (min of 1000) #
solved_states = 5   # number of eigen energy states to solve #
probe = 0           # eigen energy state you want to probe #
full = False        # do you want to solve outside of the qw? #

## Example setups ##
hBN_MoS2 = 0x01
hBN_InSe = 0x02

Selection = 0x01
if (Selection == hBN_MoS2):
     ## Example Design 1 ##
    params = {
        "EA_Cond": const.EA_MoS2_BULK,
        "m*_Cond": const.m_MoS2_BULK, 
    
        "EA_lBarrier": const.EA_hBN_MID,
        "EA_rBarrier": const.EA_hBN_MID,
        "BG_lBarrier": const.BG_hBN,
        "BG_rBarrier": const.BG_hBN,
        "lBar_Thickness": 3*const.ML_hBN,
        "rBar_Thickness": 3*const.ML_hBN,
        "Relative_Perm": const.er_hBN_BULK, 
    
        "QW_Length": 4*const.ML_MoS2,
        "m*_QW": const.m_MoS2_4L,
        "EA_QW": const.EA_MoS2_4L
    }
elif (Selection == hBN_InSe):
     ## Example Design 2 ##
    params = {
        "EA_Cond": const.EA_ITO,
        "m*_Cond": const.m_ITO, 
    
        "EA_lBarrier": const.EA_hBN_MID,
        "EA_rBarrier": const.EA_hBN_MID,
        "BG_lBarrier": const.BG_hBN,
        "BG_rBarrier": const.BG_hBN,
        "lBar_Thickness": 1*const.ML_hBN,
        "rBar_Thickness": 3*const.ML_hBN,
    
        "QW_Length": 6*const.ML_InSe,
        "m*_QW": const.m_InSe,
        "EA_QW": const.EA_InSe
    }
    
QW = qwlib.Single_Well(bias, params, resolution, full)

def QW_Mod(particle):
    return QW/const.Ha

H = Hamiltonian(particles = SingleParticle(m = params["m*_QW"]), 
				potential = QW_Mod, 
				spatial_ndim = 1, N = len(QW), extent = len(QW)*Å*0.01)

eigenstates = H.solve(max_states = solved_states)
print(eigenstates.energies)
visualization = init_visualization(eigenstates)
print(eigenstates.energies[probe])
visualization.plot_eigenstate(probe)
qwlib.Show_Potential(QW, bias, resolution)
```
